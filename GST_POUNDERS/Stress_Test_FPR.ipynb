This notebook tests whether the reduced fiducial-pair set is stable when the GST parameter vector changes.

It uses the same `smq2Q_XYICNOT` model as `GST_model.ipynb`. The reusable implementation lives in `near_minimal_fpr_reduction.py`; this notebook only calls functions from that file so it is easy to try different parameter vectors.

## Stress-Test Settings

`target` is the original pyGSTi target-model parameter vector, included only as a baseline. `random_1`, `random_2`, etc. are the sampled parameter vectors. `PARAMETER_SAMPLING_MODE` controls how random parameter vectors are generated. Use `"local_uniform"` for small bounded perturbations around the target vector, or `"global_uniform"` to draw each parameter directly between `GLOBAL_LOWER_BOUND` and `GLOBAL_UPPER_BOUND`.

```python
NUM_RANDOM_VECTORS = 10

# Options: "twirled_derivative" or "probability_jacobian_proxy".
STAGE1_METHOD = "twirled_derivative"
COMMUTANT_REL_TOL = 1e-10

# Options: "local_uniform" or "global_uniform".
PARAMETER_SAMPLING_MODE = "global_uniform"

# Used only when PARAMETER_SAMPLING_MODE = "local_uniform".
PERTURBATION_SCALE = 1e-1

# Used only when PARAMETER_SAMPLING_MODE = "global_uniform".
GLOBAL_LOWER_BOUND = -np.pi
GLOBAL_UPPER_BOUND = np.pi

# Keep True if you want the original pyGSTi target vector as a baseline row.
# Set False if you want only random vectors sampled from the bounds.
INCLUDE_TARGET_BASELINE = False

RANDOM_SEED = 1234

# Set to an integer like 100 for a faster approximate debug run.
# Keep as None for the full proxy-rank run.
MAX_STAGE1_RANK_PER_GERM = None

# Set True if you want detailed per-germ progress printed.
VERBOSE = True
```

## Run FPR At Target And Random Parameter Vectors

This can take about 25-40 seconds per parameter vector in Docker.

```python
runs, summary_df = stress_test_random_parameter_vectors(
    num_random_vectors=NUM_RANDOM_VECTORS,
    mode=PARAMETER_SAMPLING_MODE,
    perturbation_scale=PERTURBATION_SCALE,
    lower_bound=GLOBAL_LOWER_BOUND,
    upper_bound=GLOBAL_UPPER_BOUND,
    seed=RANDOM_SEED,
    include_target=INCLUDE_TARGET_BASELINE,
    stage1_method=STAGE1_METHOD,
    commutant_rel_tol=COMMUTANT_REL_TOL,
    max_stage1_rank_per_germ=MAX_STAGE1_RANK_PER_GERM,
    save_results=True,
    outdir="near_minimal_fpr_stress_results",
    verbose=VERBOSE,
)

summary_cols = [
    "label",
    "standard_circuits",
    "reduced_circuits",
    "circuits_removed",
    "reduction_fraction",
    "stage1_method",
    "stage1_global_rank",
    "stage1_proxy_global_rank",
    "total_selected_fiducial_pairs_across_germs",
    "all_stage2_ranks_complete",
    "pair_jaccard_vs_first_run",
    "sampling_mode",
    "perturbation_scale",
    "global_lower_bound",
    "global_upper_bound",
    "elapsed_seconds",
]

summary_df[summary_cols]
```

```python
# One-to-one comparison of selected fiducial pairs for one germ.
# Edit these two variables to choose which random runs and which germ to inspect.

DESIRED_RANDOM_LABELS = ["random_1", "random_2", "random_3", "random_4", "random_5", "random_6", "random_7"]
DESIRED_GERM_INDEX = 0

selected_pair_tables = []
for run in runs:
    label = run["summary"]["label"]
    if label not in DESIRED_RANDOM_LABELS:
        continue

    table = run["pair_records"].copy()
    table.insert(0, "label", label)
    selected_pair_tables.append(table)

if not selected_pair_tables:
    raise ValueError(
        f"No selected-pair data found for DESIRED_RANDOM_LABELS={DESIRED_RANDOM_LABELS}. "
        "Check that these labels exist in the current `runs` list."
    )

selected_pairs_for_desired_randoms = pd.concat(selected_pair_tables, ignore_index=True)
selected_pairs_for_desired_germ = selected_pairs_for_desired_randoms[
    selected_pairs_for_desired_randoms["germ_index"] == DESIRED_GERM_INDEX
].copy()

if selected_pairs_for_desired_germ.empty:
    raise ValueError(
        f"No selected pairs found for germ_index={DESIRED_GERM_INDEX} "
        f"inside DESIRED_RANDOM_LABELS={DESIRED_RANDOM_LABELS}."
    )

comparison_table = (
    selected_pairs_for_desired_germ
    .assign(selected=True)
    .pivot_table(
        index=["germ_index", "prep_fiducial_index", "meas_fiducial_index"],
        columns="label",
        values="selected",
        aggfunc="any",
        fill_value=False,
    )
    .reset_index()
)

# Keep the requested random-run order in the displayed columns.
comparison_table = comparison_table[
    ["germ_index", "prep_fiducial_index", "meas_fiducial_index"] + DESIRED_RANDOM_LABELS
]

for label in DESIRED_RANDOM_LABELS:
    comparison_table[label] = comparison_table[label].map({True: "yes", False: ""})

comparison_table = comparison_table.sort_values(
    ["prep_fiducial_index", "meas_fiducial_index"]
).reset_index(drop=True)

comparison_table.to_csv(
    f"near_minimal_fpr_stress_results/selected_pairs_germ_{DESIRED_GERM_INDEX}.csv",
    index=False,
)

comparison_table
```
